## Spectral simulation

```python
# Import necessary libraries
import numpy as np
import matplotlib.pyplot as plt

# Define functions for each distribution
def power_law(x, A, alpha):
    """Simple Power Law"""
    return A * x**(-alpha)

def power_law_exponential_cutoff(x, A, alpha, x_c):
    """Power Law with Exponential Cutoff"""
    return A * x**(-alpha) * np.exp(-x / x_c)

def power_law_superexponential_cutoff(x, A, alpha, x_c, beta):
    """Power Law with Superexponential Cutoff"""
    return A * x**(-alpha) * np.exp(-(x / x_c)**beta)

def broken_power_law(x, x_break, alpha1, alpha2, A):
    """Broken Power Law"""
    return np.where(x < x_break, A * (x / x_break)**(-alpha1), A * (x / x_break)**(-alpha2))

# Parameters for the distributions
A = 1.0          # Normalization
alpha = 2.1      # Power law exponent updated to 2.1
x_c = 10000.0    # Cutoff scale for the law with exponential cutoff (increased to have a quasi power-law behavior in the 5-30 GeV range)
beta = 2.5       # Superexponential cutoff rapidity
x_break = 1000   # Break point for the broken power law
alpha1 = 1.5     # Power law exponent before the break
alpha2 = 2.2     # Power law exponent after the break

# Generate synthetic data for energies from 100 MeV to 50 GeV
x = np.logspace(np.log10(100), np.log10(50000), 500)  # Energy from 100 MeV to 50 GeV (50000 MeV)

# Calculate values for each distribution
A = 1.0          # Normalization
alpha = 2.1      # Power law exponent updated to 2.1
y_powerlaw = power_law(x, A, alpha)


y_exp_cutoff = power_law_exponential_cutoff(x, A, alpha, x_c)
y_superexp_cutoff = power_law_superexponential_cutoff(x, A, alpha, x_c, beta)
y_broken_power_law = broken_power_law(x, x_break, alpha1, alpha2, A)

# Plot the distributions
plt.figure(figsize=(10, 6))

plt.plot(x, y_powerlaw, label='Simple Power Law', color='purple', linestyle='-')
plt.plot(x, y_exp_cutoff, label='Power Law with Exponential Cutoff', color='blue', linestyle='-')
plt.plot(x, y_superexp_cutoff, label='Power Law with Superexponential Cutoff', color='red', linestyle='--')
plt.plot(x, y_broken_power_law, label='Broken Power Law', color='green', linestyle='-.')

plt.xlabel('Energy (MeV)')
plt.ylabel('f(x)')
plt.title('Distributions: Power Law, Exponential Cutoff, Superexponential Cutoff, and Broken Power Law')
plt.yscale('log')
plt.xscale('log')
plt.legend()
plt.grid(True)
plt.show()

# Show the parameters used for each distribution
params = {
    "Simple Power Law": {"A": A, "alpha": alpha},
    "Power Law with Exponential Cutoff": {"A": A, "alpha": alpha, "x_c": x_c},
    "Power Law with Superexponential Cutoff": {"A": A, "alpha": alpha, "x_c": x_c, "beta": beta},
    "Broken Power Law": {"A": A, "x_break": x_break, "alpha1": alpha1, "alpha2": alpha2}
}

print(params)
```

```python
from scipy.integrate import quad
import numpy as np
from scipy.special import gamma, gammaincc

def update_integrator(eMin, eMax, index, integrator_type=3, function_type='powerlaw', Ec=None, gamma2=None, E_break=None, index1=None, index2=None, beta=None, x0=None):
    """
    Integrate a given function over specified ranges using scipy's quad integrator.

    Parameters:
    - eMin: The lower bound of the integration range.
    - eMax: The upper bound of the integration range.
    - index: The exponent parameter for the power law to integrate.
    - integrator_type: An integer indicating the precision of the integration.
    - function_type: The type of function to integrate ('powerlaw', 'superexp', 'brokenpowerlaw', 'expcutoff', 'logparabola').
    - Ec: Cutoff energy for the super exponential cutoff and exponential cutoff.
    - gamma2: Super exponential factor for the super exponential cutoff.
    - E_break: Break energy for the broken power law.
    - index1: Exponent for the power law before the break.
    - index2: Exponent for the power law after the break.
    - beta: Curvature parameter for the log-parabola.
    - x0: Reference energy for the log-parabola.

    Returns:
    - The integral of the selected function over the range [eMin, eMax].
    """

    # Define the function to integrate based on the function type
    if function_type == 'powerlaw':
        def integrand(x):
            return x ** (-index)
    elif function_type == 'superexp':
        if Ec is None or gamma2 is None:
            raise ValueError("Parameters 'Ec' and 'gamma2' must be specified for the super exponential cutoff function.")
        def integrand(x):
            return x ** (-index) * np.exp(- (x / Ec) ** gamma2)
    elif function_type == 'brokenpowerlaw':
        if E_break is None or index1 is None or index2 is None:
            raise ValueError("Parameters 'E_break', 'index1', and 'index2' must be specified for the broken power law function.")
        def integrand(x):
            return x ** (-index1) if x < E_break else x ** (-index2)
    elif function_type == 'expcutoff':
        if Ec is None:
            raise ValueError("Parameter 'Ec' must be specified for the exponential cutoff function.")
        def integrand(x):
            return x ** (-index) * np.exp(-x / Ec)
    elif function_type == 'logparabola':
        if beta is None or x0 is None:
            raise ValueError("Parameters 'beta' and 'x0' must be specified for the log-parabola function.")
        def integrand(x):
            return x ** (-index) * 10 ** (-beta * np.log10(x / x0) ** 2)
    else:
        raise ValueError("Invalid function type. Choose 'powerlaw', 'superexp', 'brokenpowerlaw', 'expcutoff', or 'logparabola'.")

    # Set relative tolerance based on integrator type
    if integrator_type == 1:
        rel_tol = 0.001
    elif integrator_type == 2:
        rel_tol = 0.000001
    elif integrator_type == 3:
        rel_tol = 0.00000001
    else:
        rel_tol = 0.000001  # Default tolerance if not specified

    # Perform the integration using scipy's quad
    integral_eMin_eMax, _ = quad(integrand, eMin, eMax, epsrel=rel_tol)

    return integral_eMin_eMax

def integrandanalytic(eMin, eMax, index, function_type='powerlaw', Ec=None, gamma2=None, E_break=None, index1=None, index2=None, beta=None, x0=None):
    """
    Compute the analytical integral of the given function type over specified ranges.

    Parameters are the same as for update_integrator.
    """

    if function_type == 'powerlaw':
        if index == 1:
            return np.log(eMax) - np.log(eMin)
        else:
            #return (eMax ** (1.0 - index) - eMin ** (1.0 - index)) / (1.0 - index)
            return (eMin ** (1.0 - index) - eMax ** (1.0 - index))
    
    elif function_type == 'expcutoff':
        if Ec is None:
            raise ValueError("Parameter 'Ec' must be specified for the exponential cutoff function.")
        # Using the incomplete gamma function for a more accurate analytical expression
        prefactor = Ec**(1 - index)
        if index >= 1:
            print(f"Warning: For index >= 1, the result may be undefined or infinite.")
            return np.nan
        try:
            result = prefactor * (gammaincc(1 - index, eMin / Ec) * gamma(1 - index) - gammaincc(1 - index, eMax / Ec) * gamma(1 - index))
        except ValueError as e:
            print(f"Error in computing incomplete gamma: {e}")
            return np.nan
        return result

    elif function_type == 'superexp':
        # No straightforward analytical solution for superexp cutoff, so return None
        return None
    
    elif function_type == 'brokenpowerlaw':
        if index1 == 1:
            integral1 = np.log(E_break) - np.log(eMin)
        else:
            integral1 = (E_break ** (1.0 - index1) - eMin ** (1.0 - index1)) / (1.0 - index1)
        
        if index2 == 1:
            integral2 = np.log(eMax) - np.log(E_break)
        else:
            integral2 = (eMax ** (1.0 - index2) - E_break ** (1.0 - index2)) / (1.0 - index2)
        
        return integral1 + integral2
    
    elif function_type == 'logparabola':
        # No straightforward analytical solution for log-parabola, so return None
        return None
    
    else:
        raise ValueError("Invalid function type. Choose 'powerlaw', 'superexp', 'brokenpowerlaw', 'expcutoff', or 'logparabola'.")

# Example usage:

# Parameters for integration
eMin = 100
eMax = 50000
index = 2.1  # Example index value for power law
integrator_type = 3

# Power law integration
result_powerlaw = update_integrator(eMin, eMax, index, integrator_type, function_type='powerlaw')
print(f"Result of the power law integration: {result_powerlaw}")

# Power law with super exponential cutoff integration
Ec = 1000  # Cutoff energy
gamma2 = 2.0  # Super exponential factor
result_superexp = update_integrator(eMin, eMax, index, integrator_type, function_type='superexp', Ec=Ec, gamma2=gamma2)
print(f"Result of the super exponential cutoff integration: {result_superexp}")

# Broken power law integration
E_break = 5000  # Break energy
index1 = 2.0  # Exponent before the break
index2 = 3.0  # Exponent after the break
result_brokenpowerlaw = update_integrator(eMin, eMax, index, integrator_type, function_type='brokenpowerlaw', E_break=E_break, index1=index1, index2=index2)
print(f"Result of the broken power law integration: {result_brokenpowerlaw}")

# Power law with exponential cutoff integration
result_expcutoff = update_integrator(eMin, eMax, index, integrator_type, function_type='expcutoff', Ec=Ec)
print(f"Result of the exponential cutoff integration: {result_expcutoff}")

# Log-parabola integration
beta = 1.5  # Curvature parameter
x0 = 1000   # Reference energy
result_logparabola = update_integrator(eMin, eMax, index, integrator_type, function_type='logparabola', beta=beta, x0=x0)
print(f"Result of the log-parabola integration: {result_logparabola}")

# Analytical expressions
result_powerlaw_analytic = integrandanalytic(eMin, eMax, index, function_type='powerlaw')
print(f"Analytical result of the power law integration: {result_powerlaw_analytic}")

result_expcutoff_analytic = integrandanalytic(eMin, eMax, index, function_type='expcutoff', Ec=Ec)
print(f"Analytical result of the exponential cutoff integration: {result_expcutoff_analytic}")

result_brokenpowerlaw_analytic = integrandanalytic(eMin, eMax, index, function_type='brokenpowerlaw', E_break=E_break, index1=index1, index2=index2)
print(f"Analytical result of the broken power law integration: {result_brokenpowerlaw_analytic}")
```
